```python
import pandas as pd
import numpy as np

forecast_results = pd.read_csv('data/processed/forecast_results.csv', parse_dates=['ds'])
forecast_results = forecast_results.rename(columns={'ds': 'date'})
metrics = pd.read_csv('data/processed/forecast_metrics.csv')
replenishment = pd.read_csv('data/processed/replenishment_table.csv')

# Simulate a "current stock" level for demo purposes (clearly noted in README as simulated)
selected = [(1,1),(5,10),(10,25)]
demo = replenishment[replenishment.apply(lambda r: (r['store'], r['item']) in selected, axis=1)].copy()
np.random.seed(42)
demo['current_stock'] = (demo['reorder_point'] * np.random.uniform(0.6, 1.3, len(demo))).round(0)
demo['needs_reorder'] = demo['current_stock'] < demo['reorder_point']

# Save as Excel (Power BI Service handles .xlsx uploads more smoothly than CSV)
with pd.ExcelWriter('data/processed/powerbi_dashboard_data.xlsx') as writer:
    forecast_results.to_excel(writer, sheet_name='forecast_vs_actual', index=False)
    metrics.to_excel(writer, sheet_name='mape_by_sku', index=False)
    demo.to_excel(writer, sheet_name='reorder_status', index=False)

print("Saved powerbi_dashboard_data.xlsx")
```
